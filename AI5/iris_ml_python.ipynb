# Iris Classification — Python

Machine learning workflow for the Iris dataset:
1. Load and summarize the data
2. Split into 80% training and 20% validation
3. Train multiple algorithms and evaluate on the validation set
4. Report the best-performing model

## Import libraries

```python
from pandas import read_csv
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
```

## Load dataset

The CSV has no header row. Column names follow the standard Iris attribute names.

```python
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/iris.csv"
names = ["sepal-length", "sepal-width", "petal-length", "petal-width", "class"]
dataset = read_csv(url, names=names)
print(f"Loaded {dataset.shape[0]} rows and {dataset.shape[1]} columns")
```

## Summarize the dataset

```python
print("Dimensions:", dataset.shape)
print("\nFirst 10 rows:")
print(dataset.head(10))
print("\nStatistical summary of numeric columns:")
print(dataset.describe())
print("\nClass distribution:")
print(dataset.groupby("class").size())
```

## Create validation dataset

Reserve 20% of the data for validation; the remaining 80% is used for training.

```python
x = dataset[["sepal-length", "sepal-width", "petal-length", "petal-width"]].values
y = dataset["class"].values

x_train, x_validation, y_train, y_validation = train_test_split(
    x, y, test_size=0.20, random_state=1, stratify=y
)

print(f"Training set: {x_train.shape[0]} samples ({x_train.shape[0] / len(y):.0%})")
print(f"Validation set: {x_validation.shape[0]} samples ({x_validation.shape[0] / len(y):.0%})")
```

## Train and compare algorithms

Each model is fit on the training set and evaluated on the held-out validation set.

```python
models = [
    ("Logistic Regression", LogisticRegression(max_iter=1000)),
    ("Linear Discriminant Analysis", LinearDiscriminantAnalysis()),
    ("k-Nearest Neighbors", KNeighborsClassifier()),
    ("Decision Tree", DecisionTreeClassifier(random_state=1)),
    ("Naive Bayes", GaussianNB()),
    ("Support Vector Machine", SVC(gamma="auto", random_state=1)),
]

results = []
for name, model in models:
    model.fit(x_train, y_train)
    predictions = model.predict(x_validation)
    accuracy = accuracy_score(y_validation, predictions)
    results.append({"algorithm": name, "accuracy": accuracy})
    print(f"{name}: {accuracy:.4f}")
```

## Best model

```python
best = max(results, key=lambda row: row["accuracy"])
print(f"Best algorithm: {best['algorithm']}")
print(f"Validation accuracy: {best['accuracy']:.4f}")
```
